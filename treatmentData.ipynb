```python
import os
import shutil
import random

# Ruta a tu dataset original
DATASET_DIR = "./archive"
OUTPUT_DIR = "dataset_split"

# Porcentaje de división
train_split = 0.7
val_split = 0.15
test_split = 0.15

# Crear carpetas de salida
for split in ["train", "val", "test"]:
    split_path = os.path.join(OUTPUT_DIR, split)
    os.makedirs(split_path, exist_ok=True)

# Recorrer cada emoción
for emotion in os.listdir(DATASET_DIR):
    emotion_path = os.path.join(DATASET_DIR, emotion)

    # Ignorar archivos sueltos
    if not os.path.isdir(emotion_path):
        continue
    
    print(f"Procesando emoción: {emotion}")

    # Obtener todas las imágenes
    images = [f for f in os.listdir(emotion_path) 
              if os.path.isfile(os.path.join(emotion_path, f))]

    # Mezclar aleatoriamente
    random.shuffle(images)

    # Calcular tamaños
    total = len(images)
    train_end = int(total * train_split)
    val_end = train_end + int(total * val_split)

    # Dividir
    train_imgs = images[:train_end]
    val_imgs = images[train_end:val_end]
    test_imgs = images[val_end:]

    # Crear carpetas de emoción dentro de cada split
    for split in ["train", "val", "test"]:
        os.makedirs(os.path.join(OUTPUT_DIR, split, emotion), exist_ok=True)

    # Mover archivos
    for img in train_imgs:
        shutil.copy(os.path.join(emotion_path, img),
                    os.path.join(OUTPUT_DIR, "train", emotion, img))

    for img in val_imgs:
        shutil.copy(os.path.join(emotion_path, img),
                    os.path.join(OUTPUT_DIR, "val", emotion, img))

    for img in test_imgs:
        shutil.copy(os.path.join(emotion_path, img),
                    os.path.join(OUTPUT_DIR, "test", emotion, img))

print("División completada.")
```
